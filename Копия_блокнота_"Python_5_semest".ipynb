```python
#4
def is_palindrome(s):
    clean = ""
    for i in s:
        if i.isalnum():
            clean += i.lower()

    left = 0
    right = len(clean) - 1

    while left < right:
        if clean[left] != clean[right]:
            return False
        left += 1
        right -= 1
    return True
print(is_palindrome("а роза упала на лапу азора"))
print(is_palindrome("abbabba"))
print(is_palindrome("race a car"))
print(is_palindrome(" "))
```

```python
#5
def search_range(nums, target):
    first = find_first(nums, target)
    second = find_second(nums, target)
    return [first, second]

def find_first(nums, target):
    left, right = 0, len(nums) - 1
    result = -1

    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            result = mid
            right = mid - 1
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

def find_second(nums, target):
    left, right = 0, len(nums) - 1
    result = -1

    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            result = mid
            left = mid + 1
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result

print(search_range([3], 3))
print()
print(search_range([0, 3, 5, 6, 7, 11, 23], 12))
print()
print(search_range([0, 3, 5, 6, 6, 6, 6, 7, 11, 23], 6))
```

```python
#6
def summa(nums, target):
    list = []
    for i in range(len(nums)):
        list.append((nums[i], i))
    list.sort()

    left = 0
    right = len(list) - 1

    while left < right:
        current = list[left][0] + list[right][0]

        if current == target:
            return [list[left][1], list[right][1]]
        elif current < target:
            left += 1
        else:
            right -= 1
    return []

print(summa([2, 5, 6, 7, 3, 0], 5))
print()
print(summa([3, 2, 4], 6))
```

```python
#7
def intersection(nums1, nums2):
    nums1.sort()
    nums2.sort()
    result = []
    last_added = None
    i = j =0
    while i < len(nums1) and j < len(nums2):
        a = nums1[i]
        b = nums2[j]
        if a < b:
            i += 1
        elif a > b:
            j += 1
        else:
            if a != last_added:
                result.append(a)
                last_added = a
            i += 1
            j += 1

    return result

print(intersection([1, 2, 2, 1], [2, 2]))
print()
print(intersection([4, 9, 5], [9, 4, 9, 8, 4]))
```

```python
import random

def identify_mages(mages):
    names = ['A', 'B', 'C', 'D', 'E']
    n = len(mages)

    # Счётчик заданных вопросов (лимит 7)
    queries_left = 7

    def ask(mage_idx, prop_fn):
        nonlocal queries_left
        if queries_left <= 0:
            return None
        queries_left -= 1
        # Задаём вопрос в форме Булоса:
        # "Ты ответишь 'zor' на вопрос, верно ли prop_fn()?"
        return mages[mage_idx](prop_fn)

    # Вероятностная модель всех допустимых конфигураций
    # До 2 случайных, минимум 1 truth, минимум 1 lie
    possible_roles = ['truth', 'lie', 'random']
    all_states = []

    for rA in possible_roles:
        for rB in possible_roles:
            for rC in possible_roles:
                for rD in possible_roles:
                    for rE in possible_roles:
                        cfg = {'A': rA, 'B': rB, 'C': rC, 'D': rD, 'E': rE}
                        r_count = sum(1 for r in cfg.values() if r == 'random')
                        t_count = sum(1 for r in cfg.values() if r == 'truth')
                        l_count = sum(1 for r in cfg.values() if r == 'lie')
                        if r_count <= 2 and t_count >= 1 and l_count >= 1:
                            all_states.append(cfg)

    # Веса конфигураций (равновероятны изначально)
    weights = {tuple(cfg.items()): 1.0 for cfg in all_states}

    # Вспомогательная функция: симуляция ответа мага i на утверждение P при конфигурации cfg
    def simulate_p(target_cfg, mage_name, prop_eval_fn):
        role = target_cfg[mage_name]
        if role == 'random':
            return 0.5  # P('zor') = 0.5
        # Для truth и lie формулировка Булоса даёт 'zor' <=> prop == True
        val = prop_eval_fn(target_cfg)
        return 1.0 if val else 0.0

    # Задаём серию из до 7 вопросов, оптимизируя информативность (Shannon entropy)
    while queries_left > 0:
        # Если осталась 1 явная конфигурация с доминирующим весом
        total_w = sum(weights.values())
        max_w = max(weights.values())
        if max_w / total_w > 0.99:
            break

        # Проверяем магов по очереди: спрашиваем про конкретные роли
        # Вопрос: "Верно ли, что маг target_idx имеет роль 'truth'?"
        # Задаём вопрос магу, который с наименьшей вероятностью random
        non_random_probs = {
            name: sum(w for cfg, w in weights.items() if dict(cfg)[name] != 'random') / total_w
            for name in names
        }
        ask_mage_name = max(non_random_probs, key=non_random_probs.get)
        ask_mage_idx = names.index(ask_mage_name)

        # Выбираем мага, о котором меньше всего определённости
        uncertainty = {
            name: sum(w for cfg, w in weights.items() if dict(cfg)[name] == 'truth') / total_w
            for name in names
        }
        # Ищем того, где вероятность ближе всего к 0.5
        target_name = min(uncertainty, key=lambda k: abs(uncertainty[k] - 0.5))

        prop = lambda cfg, t_name=target_name: cfg[t_name] == 'truth'

        # Задаём вопрос в окружение
        ans = ask(ask_mage_idx, lambda: prop)
        if ans is None:
            break

        # Байесовское обновление весов
        for cfg_tuple, w in list(weights.items()):
            cfg = dict(cfg_tuple)
            p_zor = simulate_p(cfg, ask_mage_name, prop)
            p_ans = p_zor if ans == 'zor' else (1.0 - p_zor)
            weights[cfg_tuple] = w * p_ans

    # Выбираем конфигурацию с максимальным апостериорным весом
    best_cfg = dict(max(weights.items(), key=lambda item: item[1])[0])
    return best_cfg
```

```python
def climbing_stairs_memo(n: int) -> int:
    memo = {}

    def helper(step: int) -> int:
        if step in (0, 1):
            return 1
        if step not in memo:
            memo[step] = helper(step - 1) + helper(step - 2)
        return memo[step]

    return helper(n)
```

```python
def climbing_stairs_tabulaion(n: int) -> int:
    if n in (0, 1):
        return 1

    dp = [0] * (n + 1)
    dp[0] = 1
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]
```

```python
def climbing_stairs_optimized_tabulation(n: int) -> int:
    prev2, prev1 = 1, 1
    for _ in range(2, n + 1):
        prev2, prev1 = prev1, prev2 + prev1
    return prev1
```

```python
def house_robber_memo(nums: list[int]) -> int:
    if not nums:
        return 0

    memo = {}

    def helper(i: int) -> int:
        if i == 0:
            return nums[0]
        if i == 1:
            return max(nums[0], nums[1])
        if i not in memo:
            memo[i] = max(helper(i - 1), helper(i - 2) + nums[i])
        return memo[i]

    return helper(len(nums) - 1)
```

```python
def house_robber_tabulation(nums: list[int]) -> int:
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]

    n = len(nums)
    dp = [0] * n
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, n):
        dp[i] = max(dp[i - 1], dp[i - 2] + nums[i])

    return dp[-1]
```

```python
def house_robber_optimized_tabulation(nums: list[int]) -> int:
    if not nums:
        return 0

    prev2 = 0
    prev1 = 0

    for val in nums:
        current = max(prev1, prev2 + val)
        prev2, prev1 = prev1, current

    return prev1
```

```python
def house_robber_circular(nums: list[int]) -> int:
    if not nums:
        return 0
    if len(nums) == 1:
        return nums[0]
    return max(
        house_robber_optimized_tabulation(nums[:-1]),
        house_robber_optimized_tabulation(nums[1:])
    )
```

```python
def solution1(strs):
    output = ""
    '''
    strs: list of strings
    output: string
    '''
    if not strs:
        return ""

    first_word = strs[0]
    for i in range(len(first_word)):
        char = first_word[i]
        for s in strs[1:]:
            if i >= len(s) or s[i] != char:
                return output
        output += char

    return output


print('-----Tests-----')
print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
```

```python
def solution2(arr, target):
    output = -1
    '''
    arr: list of int
    target: int
    output: int
    '''
    left, right = 0, len(arr) - 1

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    output = left
    return output


print('-----Tests-----')
print(solution2([1, 2, 3, 6, 9], 9))
print(solution2([1, 2, 3, 6, 9], 5))
print(solution2([1, 2, 3, 6, 9], 0))
print(solution2([1, 2, 3, 6, 9], 110))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

def plot_class_balance(y_train, y_val, y_test, class_names=("Норма", "Пневмония")):
    """
    Рисует три подграфика: распределение классов в train/val/test,
    а также сравнивает доли классов между выборками.
    """
    fig, axes = plt.subplots(1, 4, figsize=(20, 4.5))

    datasets = [
        ("Train", y_train, axes[0]),
        ("Validation", y_val, axes[1]),
        ("Test", y_test, axes[2]),
    ]

    for name, y, ax in datasets:
        counts = Counter(y)
        classes = sorted(counts.keys())
        labels = [class_names[c] for c in classes]
        values = [counts[c] for c in classes]
        total = sum(values)
        colors = ['#4C9F70', '#D9534F']  # зелёный — норма, красный — пневмония

        bars = ax.bar(labels, values, color=colors, edgecolor='black')
        ax.set_title(f"{name} (n = {total})", fontsize=13)
        ax.set_ylabel("Число примеров")
        for bar, v in zip(bars, values):
            pct = 100 * v / total
            ax.text(bar.get_x() + bar.get_width() / 2,
                    v + total * 0.01,
                    f"{v}\n({pct:.1f}%)",
                    ha='center', fontsize=10)

    # 4-й график: сравнение долей классов по выборкам
    ax = axes[3]
    splits = ["Train", "Val", "Test"]
    data = [y_train, y_val, y_test]
    ratios_pos = [100 * np.mean(y) for y in data]  # доля класса 1 (пневмония)

    x = np.arange(len(splits))
    width = 0.35
    ax.bar(x - width/2, [100 - r for r in ratios_pos], width,
           label=class_names[0], color='#4C9F70', edgecolor='black')
    ax.bar(x + width/2, ratios_pos, width,
           label=class_names[1], color='#D9534F', edgecolor='black')
    ax.set_xticks(x)
    ax.set_xticklabels(splits)
    ax.set_ylabel("Доля класса, %")
    ax.set_title("Сравнение долей классов", fontsize=13)
    ax.legend()
    ax.grid(axis='y', alpha=0.3)
    ax.set_ylim(0, 100)

    plt.suptitle("Баланс классов в PneumoniaMNIST", fontsize=15, y=1.02)
    plt.tight_layout()
    plt.show()

    # Печатаем числовые метрики
    print("\n" + "="*60)
    print("БАЛАНС КЛАССОВ")
    print("="*60)
    for name, y in zip(["Train", "Val", "Test"], [y_train, y_val, y_test]):
        pos = int(np.sum(y))
        neg = len(y) - pos
        ratio = pos / len(y)
        print(f"{name:6s}: всего = {len(y):5d} | "
              f"норма = {neg:5d} | пневмония = {pos:5d} | "
              f"доля пневмонии = {ratio:.2%}")


plot_class_balance(y_train, y_val, y_test)
```

```python
#1
def solution1(strs):
    output = ""
    if not strs:
        return output

    first = strs[0]
    for i, ch in enumerate(first):
        for s in strs:
            if i == len(s) or s[i] != ch:
                return first[:i]
    return first

print(solution1(["flower", "flow", "flight"]))
print(solution1(["car", "star", "wars"]))
```
