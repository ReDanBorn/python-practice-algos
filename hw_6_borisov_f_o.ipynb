```python
from sklearn.datasets import load_wine
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN, KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score

data = load_wine()
X = data.data
y = data.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

dbscan = DBSCAN(eps=1.5, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)

valid_idx = dbscan_labels != -1
db_silhouette = silhouette_score(X_scaled[valid_idx], dbscan_labels[valid_idx]) if len(np.unique(dbscan_labels[valid_idx])) > 1 else -1
db_ari = adjusted_rand_score(y, dbscan_labels)

print(f"DBSCAN Silhouette: {db_silhouette:.4f}")
print(f"DBSCAN ARI: {db_ari:.4f}")
print(f"DBSCAN clusters: {len(np.unique(dbscan_labels[valid_idx]))}")

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X_scaled)

km_silhouette = silhouette_score(X_scaled, kmeans_labels)
km_ari = adjusted_rand_score(y, kmeans_labels)

print(f"KMeans Silhouette: {km_silhouette:.4f}")
print(f"KMeans ARI: {km_ari:.4f}")
```
