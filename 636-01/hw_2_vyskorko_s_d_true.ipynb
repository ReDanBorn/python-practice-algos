### 1. Банкомат

```python
def atm(amount, limits):
  denominations = [100, 200, 500, 1000, 2000, 5000]
  if amount == 0:
    return [0,0,0,0,0,0]
  if amount % 100 != 0:
    return -1
```

```python
def atm(amount, limits):
    denominations = [100, 200, 500, 1000, 2000, 5000]
    limits_dict = dict(zip(denominations, limits))

    if amount == 0:
        return {}
    if amount % 100 != 0:
        return -1

    result = {}
    remaining = amount
    for denom in reversed(denominations):
        count = min(remaining // denom, limits_dict[denom])
        if count:
            result[denom] = count
        remaining -= count * denom

    return result if remaining == 0 else -1

print(atm(700, [10, 10, 10, 10, 10, 10]))
print(atm(350, [10, 0, 10, 10, 10, 10]))
print(atm(0, [0, 0, 0, 0, 0, 0]))
```

### 2. Подарочная карта

```python
def podarochnaya_card(prices, target):
    seen = {}
    for i, price in enumerate(prices):
        complement = target - price
        if complement in seen:
            return [seen[complement], i]
        seen[price] = i
    return []

print(podarochnaya_card([2, 7, 11, 15], 9))
print(podarochnaya_card([3, 3], 6))
```
