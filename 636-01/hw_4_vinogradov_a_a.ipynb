# Базовое ДЗ (обязательная часть)

```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_squared_error
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

california_data = fetch_california_housing(as_frame=True)
df = california_data.frame
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr(numeric_only=True), annot = True)
plt.title("Correlation Matrix")
plt.show()
```

```python
num_cols = ['MedInc', 'HouseAge', 'AveRooms', 'AveBedrms', 'Population', 'AveOccup', 'Latitude', 'Longitude']

y = df['MedHouseVal']
x = df[num_cols]
```

```python
print('Shape:', x.shape)
```

```python
x.head()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
x_train_scaled[0]
```

```python
models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(),
    'Lasso': Lasso(),
    'RandomForestRegressor': RandomForestRegressor(random_state=42),
}

default_acc = {}
for name, model in models.items():
    model.fit(x_train_scaled, y_train)
    default_acc[name] = mean_squared_error(y_test, model.predict(x_test_scaled))
    print(name, '->', round(default_acc[name], 4))

# Оценка сложности решения по времени и памяти (худший случай):
# n — количество объектов, p — число признаков (здесь p = 8),
# T — число деревьев в RandomForestRegressor (по умолчанию T = 100),
# I — число итераций оптимизации Lasso.
# Разделение и масштабирование данных: время O(n*p), память O(n*p).
# LinearRegression / Ridge: примерно O(n*p^2 + p^3) при n >= p;
#   точные затраты зависят от численного метода в sklearn.
# Lasso: O(I*n*p) для координатных проходов по признакам.
# RandomForestRegressor: грубая верхняя оценка O(T*p*n^2*log(n))
#   на обучение при очень неудачном (несбалансированном) росте деревьев.
# Общая оценка в худшем случае при фиксированных p, T, I:
#   время — O(n^2*log(n)); память — O(n) (данные и узлы деревьев).
# Это верхняя оценка: на практике случайный лес обычно работает быстрее.
```
